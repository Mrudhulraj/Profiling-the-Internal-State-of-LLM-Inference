```python
import torch
import time
import numpy as np
from pathlib import Path
from litgpt import LLM
from datasets import load_dataset

# --- Configuration ---
# LitGPT downloads models to ./checkpoints/ by default
MODEL_REPO = "meta-llama/Llama-3.2-1B"
CHECKPOINT_DIR = Path("checkpoints") / MODEL_REPO

DATASET_ID = "tatsu-lab/alpaca"
NUM_SAMPLES = 100
MAX_NEW_TOKENS = 50
LAYER_ID = 4
HEAD_ID = 0

def run_litgpt_benchmark():
    print(f"\n=== LitGPT Benchmark (Alpaca Dataset) ===")

    # 1. Check if model exists
    if not CHECKPOINT_DIR.exists():
        raise FileNotFoundError(f"Model not found at {CHECKPOINT_DIR}. Did you run Step 3?")

    # 2. Load Data
    print(f"Loading {DATASET_ID}...")
    ds = load_dataset(DATASET_ID, split="train")
    prompts = [item['instruction'] for item in ds.select(range(NUM_SAMPLES))]

    # 3. Load Model
    print(f"Loading Model from {CHECKPOINT_DIR}...")
    # LitGPT's LLM.load can take the path directly
    llm = LLM.load(CHECKPOINT_DIR)
    llm.model.eval()

    # 4. Define Hook for Fused QKV
    captured_data = {"Q": [], "K": [], "V": []}

    def fused_hook(module, input, output):
        # LitGPT output shape: [batch, seq, (n_head + 2*n_query_groups) * head_size]
        config = llm.model.config
        head_size = config.head_size
        n_head = config.n_head
        n_query_groups = config.n_query_groups

        # Calculate split points
        q_end = n_head * head_size
        k_end = q_end + (n_query_groups * head_size)

        # Slice Q (we focus on Q for the benchmark metric)
        q = output[..., :q_end]

        # Reshape Q: [batch, seq, n_head, head_size]
        bsz, seq, _ = q.shape
        q_reshaped = q.view(bsz, seq, n_head, head_size)

        # Capture specific head (last token only)
        vector = q_reshaped[:, -1:, HEAD_ID, :].detach().cpu()
        captured_data["Q"].append(vector)

    # 5. Register Hook
    # Path: model -> transformer -> h (layers) -> attn (CausalSelfAttention module)
    target_layer = llm.model.transformer.h[LAYER_ID].attn
    handle = target_layer.register_forward_hook(fused_hook)

    # 6. Run Inference Loop
    latencies = []
    print(f"Starting Inference on {NUM_SAMPLES} prompts...")

    # Warmup (optional but recommended for benchmarks)
    print("Warming up...")
    llm.generate("Hello", max_new_tokens=5)

    for i, prompt in enumerate(prompts):
        torch.cuda.synchronize()
        t0 = time.time()

        # LitGPT generate handles tokenization internally
        _ = llm.generate(prompt, max_new_tokens=MAX_NEW_TOKENS)

        torch.cuda.synchronize()
        t1 = time.time()
        latencies.append(t1 - t0)
        print(f"Sample {i+1}/{NUM_SAMPLES}: {t1 - t0:.4f}s")

    # Cleanup
    handle.remove()

    # Results
    avg_latency = np.mean(latencies)
    print(f"\n--- LitGPT Results ---")
    print(f"Average Latency: {avg_latency:.4f}s per prompt")
    print(f"Captured Vector Count (Q): {len(captured_data['Q'])}")
    if len(captured_data['Q']) > 0:
        print(f"Vector Shape (Single Step): {captured_data['Q'][0].shape}")

    return avg_latency

if __name__ == "__main__":
    run_litgpt_benchmark()
```

```python
import torch
import time
import numpy as np
from datasets import load_dataset
from transformers import AutoModelForCausalLM, AutoTokenizer
from litgpt import LLM

# Clear CUDA cache before starting, just in case previous runs left memory allocated
torch.cuda.empty_cache()

# --- Configuration ---
MODEL_ID = "meta-llama/Llama-3.2-1B"
DATASET_ID = "tatsu-lab/alpaca"
NUM_SAMPLES = 5        # Reduced count slightly as float32 is slower
LAYER_ID = 2
HEAD_ID = 4
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# 1. Force Deterministic Algorithms
torch.manual_seed(42)
torch.use_deterministic_algorithms(True, warn_only=True)
torch.backends.cudnn.deterministic = True
torch.backends.cuda.matmul.allow_tf32 = False # Disable TF32 optimizations

print(f"--- Setting up PURE FLOAT32 Benchmark on {DEVICE} ---")

# 2. Load Data & Tokenizer
print(f"Loading {DATASET_ID}...")
ds = load_dataset(DATASET_ID, split="train")
tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)
samples = [item['instruction'] + " " + item['input'] for item in ds.select(range(NUM_SAMPLES))]

results = {
    "hf_latencies": [],
    "lit_latencies": [],
    "cosine_similarities": [],
    "mse_losses": []
}
hf_all_activations = [] # To store activations for all samples on CPU
lit_all_activations = [] # To store activations for all samples on CPU

# ==========================================
# PART A: HUGGING FACE (FLOAT 32) - Processing all samples
# ==========================================
print("Loading Hugging Face (Float32)...")
hf_model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    torch_dtype=torch.float32,      # FORCE FLOAT32
    attn_implementation="eager"     # FORCE VANILLA ATTENTION (No Flash Attn)
).to(DEVICE)
hf_model.eval()

# Define and register HF hook for this phase
hf_current_activation = {} # Temporarily store activation for current sample
def get_hf_hook(name):
    def hook(module, input, output):
        bsz, seq, _ = output.shape
        num_heads = hf_model.config.num_attention_heads
        head_dim = output.shape[-1] // num_heads
        reshaped = output.view(bsz, seq, num_heads, head_dim)
        hf_current_activation[name] = reshaped[0, :, HEAD_ID, :].detach().cpu() # Store on CPU
    return hook
hf_hook_handle = hf_model.model.layers[LAYER_ID].self_attn.q_proj.register_forward_hook(get_hf_hook("Q"))

print(f"Running Hugging Face passes for {NUM_SAMPLES} samples...")
for i, text in enumerate(samples):
    inputs = tokenizer(text, return_tensors="pt").to(DEVICE)
    input_ids = inputs["input_ids"]

    torch.cuda.synchronize()
    t0 = time.time()
    with torch.no_grad():
        _ = hf_model(input_ids)
    torch.cuda.synchronize()
    results["hf_latencies"].append(time.time() - t0)
    hf_all_activations.append(hf_current_activation["Q"]) # Save to list

hf_hook_handle.remove()
del hf_model
torch.cuda.empty_cache()
print("Hugging Face processing complete and model unloaded.")


# ==========================================
# PART B: LITGPT (FLOAT 32) - Processing all samples
# ==========================================
print("\nLoading LitGPT (Float32)...")
lit_llm = LLM.load(MODEL_ID)
lit_model = lit_llm.model.to(DEVICE)

# FORCE FLOAT32 CASTING
# LitGPT loads in bf16 by default for Llama, so we manually cast it up.
lit_model = lit_model.float()
lit_model.eval()

# Define and register LitGPT hook for this phase
lit_current_activation = {} # Temporarily store activation for current sample
def lit_fused_hook(module, input, output):
    config = lit_model.config
    head_size = config.head_size
    n_head = config.n_head
    q_end = n_head * head_size
    q = output[..., :q_end].view(1, -1, n_head, head_size)
    lit_current_activation["Q"] = q[0, :, HEAD_ID, :].detach().cpu() # Store on CPU
lit_hook_handle = lit_model.transformer.h[LAYER_ID].attn.register_forward_hook(lit_fused_hook)

print(f"Running LitGPT passes for {NUM_SAMPLES} samples...")
for i, text in enumerate(samples):
    inputs = tokenizer(text, return_tensors="pt").to(DEVICE) # Still need to tokenize and send to GPU
    input_ids = inputs["input_ids"]

    torch.cuda.synchronize()
    t0 = time.time()
    with torch.no_grad():
        _ = lit_model(input_ids)
    torch.cuda.synchronize()
    results["lit_latencies"].append(time.time() - t0)
    lit_all_activations.append(lit_current_activation["Q"]) # Save to list

lit_hook_handle.remove()
del lit_model
torch.cuda.empty_cache()
print("LitGPT processing complete and model unloaded.")

# ==========================================
# PART C: STATISTICAL COMPARISON - Post-processing
# ==========================================
print("\n" + "="*40)
print("PURE FLOAT32 STATISTICS")
print("="*40)

# Calculate comparisons *after* both models have run and unloaded
for i in range(NUM_SAMPLES):
    vec_hf = hf_all_activations[i]
    vec_lit = lit_all_activations[i]

    cos = torch.nn.functional.cosine_similarity(vec_hf.flatten(), vec_lit.flatten(), dim=0)
    mse = torch.nn.functional.mse_loss(vec_hf, vec_lit)

    results["cosine_similarities"].append(cos.item())
    results["mse_losses"].append(mse.item())

print(f"Avg HF Latency:     {np.mean(results['hf_latencies']):.5f} s")
print(f"Avg LitGPT Latency: {np.mean(results['lit_latencies']):.5f} s")
print("-" * 40)
print(f"Avg Cosine Sim:     {np.mean(results['cosine_similarities']):.6f}")
print(f"Avg MSE Loss:       {np.mean(results['mse_losses']):.10f}")

if np.mean(results['cosine_similarities']) > 0.9999:
    print("\n✅ PERFECT MATCH: Implementation differences eliminated.")
else:
    print("\n⚠️ MINOR DRIFT: Likely due to RoPE implementation differences (sin/cos table precision).")
```

```python
import torch
import os
import numpy as np
from transformers import AutoModelForCausalLM, AutoTokenizer
from litgpt import LLM

# Configuration
MODEL_ID = "meta-llama/Llama-3.2-1B"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print(f"--- STARTING AUTOPSY ON {DEVICE} ---")

# =========================================================
# 1. LOAD HUGGING FACE (Ground Truth)
# =========================================================
print("\n[1/4] Loading Hugging Face Model...")
hf_model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID, torch_dtype=torch.float32
).to(DEVICE)
hf_model.eval()

# Get Embedding Weights
hf_emb_weight = hf_model.model.embed_tokens.weight.detach().cpu()
# Get Layer 0 Attention Weights (Query Projection only)
hf_q_weight = hf_model.model.layers[0].self_attn.q_proj.weight.detach().cpu()

print(f"   HF Embeddings Shape: {hf_emb_weight.shape}")
print(f"   HF Query Weight Shape: {hf_q_weight.shape}")

# Generate a reference embedding for a specific token
# Token ID 1 is usually '<s>' or start token
test_token_id = torch.tensor([[1]]).to(DEVICE)
with torch.no_grad():
    hf_token_out = hf_model.model.embed_tokens(test_token_id).cpu()

# Clear HF to save memory
del hf_model
torch.cuda.empty_cache()


# =========================================================
# 2. LOAD LITGPT
# =========================================================
print("\n[2/4] Loading LitGPT Model...")
lit_llm = LLM.load(MODEL_ID)
lit_model = lit_llm.model.to(DEVICE).float().eval()

# Get Embedding Weights
lit_emb_weight = lit_model.transformer.wte.weight.detach().cpu()

# Get Layer 0 Attention Weights (Fused QKV)
# LitGPT fuses Q, K, V into one matrix. We need to slice it.
lit_fused_weight = lit_model.transformer.h[0].attn.qkv.weight.detach().cpu()
config = lit_model.config
# Slice out just the Q part (First chunk)
q_size = config.n_head * config.head_size
lit_q_weight = lit_fused_weight[:q_size, :]

print(f"   Lit Embeddings Shape: {lit_emb_weight.shape}")
print(f"   Lit Fused Weight Shape: {lit_fused_weight.shape}")
print(f"   Lit Sliced Q Weight Shape: {lit_q_weight.shape}")

with torch.no_grad():
    lit_token_out = lit_model.transformer.wte(test_token_id).cpu()

del lit_model
torch.cuda.empty_cache()


# =========================================================
# 3. DIAGNOSTIC COMPARISON
# =========================================================
print("\n[3/4] ANALYZING DIFFERENCES...")

# TEST A: DO THE EMBEDDING WEIGHTS MATCH?
# If this fails, you are loading different models (or random init).
emb_sim = torch.nn.functional.cosine_similarity(hf_emb_weight.flatten(), lit_emb_weight.flatten(), dim=0)
print(f"A. Embedding Weights Similarity:  {emb_sim.item():.6f}")

# TEST B: DO THE TOKEN OUTPUTS MATCH?
# Even if weights match, some models scale embeddings by sqrt(dim).
token_diff = (hf_token_out - lit_token_out).abs().max()
print(f"B. Token Output Max Difference:   {token_diff.item():.6f}")

# TEST C: DO THE Q-PROJECTION WEIGHTS MATCH?
# If embeddings match but this fails, LitGPT is permuting weights for RoPE.
q_sim = torch.nn.functional.cosine_similarity(hf_q_weight.flatten(), lit_q_weight.flatten(), dim=0)
print(f"C. Query Layer Weights Similarity: {q_sim.item():.6f}")


# =========================================================
# 4. CONCLUSION
# =========================================================
print("\n[4/4] DIAGNOSIS")
if emb_sim < 0.99:
    print("❌ CRITICAL FAILURE: Embedding weights do not match.")
    print("   Likely Cause: LitGPT initialized a random model instead of loading checkpoint.")
    print("   Solution: Check your 'litgpt download' path.")
elif q_sim < 0.99:
    print("⚠️ WEIGHT MISMATCH: Embeddings are good, but Attention Layers differ.")
    print("   Likely Cause: LitGPT permutes weights for RoPE optimization.")
    print("   This explains why your extracted vectors don't match.")
else:
    print("✅ EVERYTHING MATCHES: The weights are identical.")
    print("   If vectors still differ, the issue is strictly in the Forward Pass code (RoPE application).")
```

```python
import torch
import numpy as np
from transformers import AutoModelForCausalLM, AutoTokenizer
from litgpt import LLM

# Configuration
MODEL_ID = "meta-llama/Llama-3.2-1B"
DEVICE = "cuda"

print(f"--- STARTING SINGLE-TOKEN TRACE ON {DEVICE} ---")
torch.set_printoptions(linewidth=200, precision=4)

# 1. Setup Data
tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)
# We use "The" (This will likely result in [BOS, "The"])
inputs = tokenizer("The", return_tensors="pt")
input_ids = inputs["input_ids"].to(DEVICE)

print(f"Input Shape: {input_ids.shape}")
print(f"Input IDs:   {input_ids.tolist()}")

# =========================================================
# PHASE 1: HUGGING FACE TRACE
# =========================================================
print("\n>>> Tracing Hugging Face...")
hf_model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    torch_dtype=torch.float32,
    attn_implementation="eager"
).to(DEVICE)
hf_model.eval()

hf_trace = {}

# 1. Embedding Output
emb_out = hf_model.model.embed_tokens(input_ids)
hf_trace["embedding"] = emb_out.detach().cpu().float()

# 2. Layer 0 RMSNorm Output (The input to Attention)
norm = hf_model.model.layers[0].input_layernorm
norm_out = norm(emb_out)
hf_trace["rms_norm"] = norm_out.detach().cpu().float()

# 3. Layer 0 Q-Proj Output (Linear Projection ONLY)
# Note: HF Q-proj is often just a Linear layer. We run it manually.
q_proj = hf_model.model.layers[0].self_attn.q_proj
q_out = q_proj(norm_out)
hf_trace["q_out"] = q_out.detach().cpu().float()

del hf_model
torch.cuda.empty_cache()


# =========================================================
# PHASE 2: LITGPT TRACE
# =========================================================
print("\n>>> Tracing LitGPT...")
lit_llm = LLM.load(MODEL_ID)
lit_model = lit_llm.model.to(DEVICE).float().eval()

lit_trace = {}

# 1. Embedding Output
lit_emb = lit_model.transformer.wte(input_ids)
lit_trace["embedding"] = lit_emb.detach().cpu().float()

# 2. Layer 0 RMSNorm Output
# LitGPT usually names this 'norm_1' inside the block
block = lit_model.transformer.h[0]
lit_norm = block.norm_1(lit_emb)
lit_trace["rms_norm"] = lit_norm.detach().cpu().float()

# 3. Layer 0 Q-Proj Output (Manual Slice)
config = lit_model.config
head_dim = config.head_size
n_head = config.n_head
q_dim = n_head * head_dim

# Get the fused weight [3072, 2048]
# FIXED: Use .qkv instead of .attn
fused_weight = block.attn.qkv.weight
# Slice out Q [2048, 2048] - The first chunk
q_weight = fused_weight[:q_dim, :]

# Manual Linear: (Input @ Weight.T)
# We use F.linear to ensure exact PyTorch math
lit_q_out = torch.nn.functional.linear(lit_norm, q_weight)
lit_trace["q_out"] = lit_q_out.detach().cpu().float()

del lit_model
torch.cuda.empty_cache()


# =========================================================
# PHASE 3: COMPARISON REPORT
# =========================================================
print("\n" + "="*40)
print("TRACE REPORT")
print("="*40)

steps = ["embedding", "rms_norm", "q_out"]

for step in steps:
    t_hf = hf_trace[step]
    t_lit = lit_trace[step]

    # Flatten for global comparison
    sim = torch.nn.functional.cosine_similarity(t_hf.flatten(), t_lit.flatten(), dim=0).item()
    mse = torch.nn.functional.mse_loss(t_hf, t_lit).item()
    max_diff = (t_hf - t_lit).abs().max().item()

    status = "✅ MATCH" if sim > 0.9999 else "❌ FAIL"
    print(f"{step.upper():<12} | Sim: {sim:.6f} | MaxDiff: {max_diff:.6f} | {status}")

print("-" * 40)
if hf_trace["q_out"].shape == lit_trace["q_out"].shape:
    print("Shapes Match: Yes")
else:
    print(f"Shapes Mismatch: HF={hf_trace['q_out'].shape} vs Lit={lit_trace['q_out'].shape}")
```

```python
import os, json, time, math, statistics, pathlib, urllib.request
from dataclasses import dataclass
from typing import Dict, List, Tuple, Optional

import numpy as np
import torch
from tqdm import tqdm

# ----------------------------
# Config
# ----------------------------
MODEL_ID = "meta-llama/Llama-3.2-1B"  # base model
DEVICE = "cuda"
DTYPE = torch.float16  # T4-friendly

# "Industry standard" prompts: MT-Bench questions (we use turns[0] as single-turn prompts)
# Source: FastChat MT-bench file on GitHub :contentReference[oaicite:4]{index=4}
MT_BENCH_RAW_URL = "https://raw.githubusercontent.com/lm-sys/FastChat/main/fastchat/llm_judge/data/mt_bench/question.jsonl"

# What to capture
LAYERS_TO_CAPTURE = [0, 6, 12]              # change as needed
HEADS_TO_CAPTURE = [0, 3, 7]                # query head indices
TOKEN_POSITIONS = [0, 31, 63]               # token indices within prompt (will clamp if shorter)

# Benchmarking
MAX_PROMPTS = 80    # MT-bench has 80 questions; keep <=80 for full set
WARMUP = 10
TRUNCATE_TO_TOKENS = 512  # truncate prompts to max length for stable compare
BATCH_SIZE = 1            # keep 1 for simplest apples-to-apples on Colab
SAVE_RAW_SLICES_FOR_FIRST_N = 3  # set 0 to disable raw tensor saving

OUT_DIR = pathlib.Path("qkv_benchmark_out")
OUT_DIR.mkdir(exist_ok=True)

# ----------------------------
# Utilities
# ----------------------------
def ensure_hf_login():
    # Use env var HF_TOKEN if present
    token = os.environ.get("HF_TOKEN") or os.environ.get("HUGGINGFACE_TOKEN")
    if token:
        from huggingface_hub import login
        login(token=token, add_to_git_credential=False)
    else:
        print("WARNING: HF_TOKEN not set. If the model is gated, downloads will fail.")

def download_mtbench(path: pathlib.Path):
    if path.exists():
        return
    print(f"Downloading MT-Bench prompts to {path} ...")
    urllib.request.urlretrieve(MT_BENCH_RAW_URL, path)

def load_mtbench_first_turns(path: pathlib.Path, max_prompts: int) -> List[str]:
    prompts = []
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            ex = json.loads(line)
            prompts.append(ex["turns"][0])  # base model: use first user turn
            if len(prompts) >= max_prompts:
                break
    return prompts

def cuda_time_ms(fn):
    torch.cuda.synchronize()
    start = torch.cuda.Event(enable_timing=True)
    end = torch.cuda.Event(enable_timing=True)
    start.record()
    fn()
    end.record()
    torch.cuda.synchronize()
    return float(start.elapsed_time(end))

def percentile(xs, p):
    if not xs:
        return float("nan")
    xs_sorted = sorted(xs)
    k = (len(xs_sorted) - 1) * (p / 100.0)
    f = math.floor(k)
    c = math.ceil(k)
    if f == c:
        return float(xs_sorted[int(k)])
    d0 = xs_sorted[f] * (c - k)
    d1 = xs_sorted[c] * (k - f)
    return float(d0 + d1)

# ----------------------------
# HF Transformers runner
# ----------------------------
def hf_load():
    from transformers import AutoTokenizer, AutoModelForCausalLM
    tok = AutoTokenizer.from_pretrained(MODEL_ID, use_fast=True)
    model = AutoModelForCausalLM.from_pretrained(
        MODEL_ID,
        torch_dtype=DTYPE,
        device_map="cuda",
    ).eval()
    return tok, model

def hf_attach_qkv_hooks(model, layers_to_capture, heads_to_capture, token_positions, store_dict, save_raw_for_first_n=0):
    handles = []
    saved_count = {"n": 0}

    def get_attn_dims(attn_module):
        cfg = getattr(attn_module, "config", None) or getattr(attn_module, "args", None)

        # heads
        n_heads = getattr(attn_module, "num_heads", None)
        if n_heads is None and cfg is not None:
            n_heads = getattr(cfg, "num_attention_heads", None)
        if n_heads is None:
            # fallback: infer from q_proj output if possible
            # q_proj output dim = n_heads * head_dim
            # if head_dim known, can infer; otherwise we’ll infer head_dim later
            n_heads = getattr(cfg, "n_head", None) if cfg is not None else None

        # kv heads (GQA)
        n_kv = getattr(attn_module, "num_key_value_heads", None)
        if n_kv is None and cfg is not None:
            n_kv = getattr(cfg, "num_key_value_heads", None)
        if n_kv is None:
            n_kv = n_heads

        # head_dim
        head_dim = getattr(attn_module, "head_dim", None)
        if head_dim is None and cfg is not None:
            head_dim = getattr(cfg, "head_dim", None)
        if head_dim is None and cfg is not None:
            hidden = getattr(cfg, "hidden_size", None)
            if hidden is not None and n_heads is not None:
                head_dim = hidden // n_heads

        # final fallback: infer head_dim from q_proj output dim
        if head_dim is None and n_heads is not None:
            q_out = attn_module.q_proj.out_features
            head_dim = q_out // n_heads

        if n_heads is None or n_kv is None or head_dim is None:
            raise RuntimeError(
                f"Could not infer attention dims. n_heads={n_heads}, n_kv={n_kv}, head_dim={head_dim}. "
                f"Config={type(cfg).__name__ if cfg is not None else None}"
            )

        return int(n_heads), int(n_kv), int(head_dim)

    def make_pre_hook(layer_idx):
        def pre_hook(attn_module, inputs, kwargs):
            hidden = None
            if inputs and len(inputs) > 0:
                hidden = inputs[0]
            elif kwargs and "hidden_states" in kwargs:
                hidden = kwargs["hidden_states"]
            if hidden is None:
                return

            B, T, _ = hidden.shape

            # Projections
            q = attn_module.q_proj(hidden)
            k = attn_module.k_proj(hidden)
            v = attn_module.v_proj(hidden)

            n_heads, n_kv, head_dim = get_attn_dims(attn_module)

            q = q.view(B, T, n_heads, head_dim).transpose(1, 2)  # [B,H,T,D]
            k = k.view(B, T, n_kv, head_dim).transpose(1, 2)     # [B,Hkv,T,D]
            v = v.view(B, T, n_kv, head_dim).transpose(1, 2)

            # Map query heads -> kv heads for GQA
            group = max(1, n_heads // n_kv)
            kv_heads = [h // group for h in heads_to_capture]

            # Clamp token positions
            toks = [min(tp, T - 1) for tp in token_positions if T > 0]
            if not toks:
                return

            # Select slices
            q_sel = q[:, heads_to_capture, :, :][:, :, toks, :]
            k_sel = k[:, kv_heads, :, :][:, :, toks, :]
            v_sel = v[:, kv_heads, :, :][:, :, toks, :]

            stats = {
                "q_l2": float(torch.linalg.norm(q_sel).detach().cpu()),
                "k_l2": float(torch.linalg.norm(k_sel).detach().cpu()),
                "v_l2": float(torch.linalg.norm(v_sel).detach().cpu()),
                "shape_q": tuple(q_sel.shape),
                "shape_k": tuple(k_sel.shape),
                "shape_v": tuple(v_sel.shape),
                "T": int(T),
                "n_heads": int(n_heads),
                "n_kv": int(n_kv),
                "head_dim": int(head_dim),
            }
            store_dict[layer_idx] = stats

            if save_raw_for_first_n and saved_count["n"] < save_raw_for_first_n:
                saved_count["n"] += 1
                q_cpu = q_sel.detach().float().cpu().numpy()
                k_cpu = k_sel.detach().float().cpu().numpy()
                v_cpu = v_sel.detach().float().cpu().numpy()
                np.savez_compressed(
                    OUT_DIR / f"hf_qkv_layer{layer_idx}_sample{saved_count['n']}.npz",
                    q=q_cpu, k=k_cpu, v=v_cpu
                )

        return pre_hook

    for li in layers_to_capture:
        attn = model.model.layers[li].self_attn
        handles.append(attn.register_forward_pre_hook(make_pre_hook(li), with_kwargs=True))

    return handles



@torch.inference_mode()
def hf_prefill(tok, model, prompts: List[str], capture: bool) -> Tuple[List[float], List[int], List[Dict]]:
    ms_list, tok_list, cap_list = [], [], []
    for p in prompts:
        inputs = tok(p, return_tensors="pt", truncation=True, max_length=TRUNCATE_TO_TOKENS).to(DEVICE)
        n_tokens = int(inputs["input_ids"].shape[1])

        # run forward
        cap_snapshot = {}
        def run():
            _ = model(**inputs)

        ms = cuda_time_ms(run)
        ms_list.append(ms)
        tok_list.append(n_tokens)

        if capture:
            # hooks already populated global store; we capture a copy here
            cap_list.append(dict(HF_CAPTURES))
        else:
            cap_list.append({})
    return ms_list, tok_list, cap_list

# ----------------------------
# LitGPT runner
# ----------------------------
def download_hf_repo_to_dir(repo_id: str, local_dir: str):
    from huggingface_hub import snapshot_download
    token = os.environ.get("HF_TOKEN") or os.environ.get("HUGGINGFACE_TOKEN")
    return snapshot_download(
        repo_id=repo_id,
        local_dir=local_dir,
        local_dir_use_symlinks=False,
        token=token,
    )

def litgpt_load():
    import os
    from huggingface_hub import login
    from litgpt import LLM

    token = os.environ.get("HF_TOKEN") or os.environ.get("HUGGINGFACE_TOKEN")
    if token:
        login(token=token, add_to_git_credential=False)

    llm = LLM.load(MODEL_ID, distribute=None)

    if getattr(llm, "model", None) is None:
        # fallback: force snapshot download then load from local dir
        local_ckpt = "litgpt_llama32_1b_ckpt"
        print("LitGPT load from HF repo ID failed; downloading snapshot to", local_ckpt)
        download_hf_repo_to_dir(MODEL_ID, local_ckpt)

        # Try loading from the local folder
        llm = LLM.load(local_ckpt, distribute=None)

    if getattr(llm, "model", None) is None:
        raise RuntimeError("LitGPT still returned llm.model=None after local snapshot download.")

    llm.model.eval()
    return llm


def litgpt_find_attention_module(block) -> Optional[torch.nn.Module]:
    """
    Robustly find an attention-like module inside a LitGPT block.
    We look for a submodule that has either:
      - q_proj/k_proj/v_proj
      - or a combined qkv (common in minimal transformer impls)
    """
    for m in block.modules():
        if hasattr(m, "q_proj") and hasattr(m, "k_proj") and hasattr(m, "v_proj"):
            return m
        if hasattr(m, "qkv") and isinstance(getattr(m, "qkv"), torch.nn.Module):
            return m
    return None

def litgpt_get_heads_config(llm) -> Tuple[int, int, int]:
    """
    Try to infer n_heads, n_kv_heads, head_dim from LitGPT config.
    Falls back to computing head_dim from model width.
    """
    cfg = getattr(llm.model, "config", None)
    if cfg is None:
        raise RuntimeError("LitGPT model config not found.")

    # Common names across configs
    n_heads = int(getattr(cfg, "n_head", getattr(cfg, "n_heads", None)))
    n_embd = int(getattr(cfg, "n_embd", getattr(cfg, "hidden_size", None)))

    # GQA: might be n_query_groups or n_kv_heads
    n_kv = getattr(cfg, "n_query_groups", None)
    if n_kv is None:
        n_kv = getattr(cfg, "n_kv_heads", None)
    if n_kv is None:
        # assume MHA
        n_kv = n_heads
    n_kv = int(n_kv)

    head_dim = n_embd // n_heads
    return n_heads, n_kv, head_dim

def litgpt_attach_qkv_hooks(llm, layers_to_capture, heads_to_capture, token_positions, store_dict, save_raw_for_first_n=0):
    handles = []
    saved_count = {"n": 0}
    n_heads, n_kv, head_dim = litgpt_get_heads_config(llm)

    def make_hook(layer_idx, attn_module):
        def hook(module, inputs, output):
            hidden = inputs[0]  # [B,T,C]
            B, T, C = hidden.shape

            # Compute q/k/v (support both split proj and combined qkv)
            if hasattr(attn_module, "q_proj") and hasattr(attn_module, "k_proj") and hasattr(attn_module, "v_proj"):
                q = attn_module.q_proj(hidden)
                k = attn_module.k_proj(hidden)
                v = attn_module.v_proj(hidden)
            elif hasattr(attn_module, "qkv"):
                qkv = attn_module.qkv(hidden)  # [B,T, 3*C] typically
                q, k, v = torch.split(qkv, C, dim=-1)
            else:
                return

            q = q.view(B, T, n_heads, head_dim).transpose(1, 2)  # [B,H,T,D]
            k = k.view(B, T, n_kv, head_dim).transpose(1, 2)     # [B,Hkv,T,D]
            v = v.view(B, T, n_kv, head_dim).transpose(1, 2)

            group = n_heads // n_kv
            kv_heads = [h // group for h in heads_to_capture]

            toks = [min(tp, T - 1) for tp in token_positions if T > 0]
            if not toks:
                return

            q_sel = q[:, heads_to_capture, :, :][:, :, toks, :]
            k_sel = k[:, kv_heads, :, :][:, :, toks, :]
            v_sel = v[:, kv_heads, :, :][:, :, toks, :]

            stats = {
                "q_l2": float(torch.linalg.norm(q_sel).detach().cpu()),
                "k_l2": float(torch.linalg.norm(k_sel).detach().cpu()),
                "v_l2": float(torch.linalg.norm(v_sel).detach().cpu()),
                "shape_q": tuple(q_sel.shape),
                "shape_k": tuple(k_sel.shape),
                "shape_v": tuple(v_sel.shape),
                "T": int(T),
                "n_heads": int(n_heads),
                "n_kv": int(n_kv),
            }
            store_dict[layer_idx] = stats

            if save_raw_for_first_n and saved_count["n"] < save_raw_for_first_n:
                saved_count["n"] += 1
                q_cpu = q_sel.detach().float().cpu().numpy()
                k_cpu = k_sel.detach().float().cpu().numpy()
                v_cpu = v_sel.detach().float().cpu().numpy()
                np.savez_compressed(OUT_DIR / f"litgpt_qkv_layer{layer_idx}_sample{saved_count['n']}.npz", q=q_cpu, k=k_cpu, v=v_cpu)

        return hook

    # LitGPT blocks are typically under llm.model.transformer.h (ModuleList)
    blocks = llm.model.transformer.h
    for li in layers_to_capture:
        block = blocks[li]
        attn = litgpt_find_attention_module(block)
        if attn is None:
            raise RuntimeError(f"Could not find attention module in LitGPT block {li}. "
                               f"Try printing llm.model.transformer.h[{li}] to inspect names.")
        handles.append(attn.register_forward_hook(make_hook(li, attn)))

    return handles

@torch.inference_mode()
def litgpt_prefill(llm, prompts: List[str], capture: bool) -> Tuple[List[float], List[int], List[Dict]]:
    ms_list, tok_list, cap_list = [], [], []
    for p in prompts:
        # LitGPT tokenizer: encode returns 1D tensor of token ids
        ids = llm.tokenizer.encode(p, device=DEVICE)
        if ids.numel() > TRUNCATE_TO_TOKENS:
            ids = ids[:TRUNCATE_TO_TOKENS]
        input_ids = ids.unsqueeze(0)  # [1,T]
        n_tokens = int(input_ids.shape[1])

        def run():
            _ = llm(input_ids=input_ids)

        ms = cuda_time_ms(run)
        ms_list.append(ms)
        tok_list.append(n_tokens)

        if capture:
            cap_list.append(dict(LIT_CAPTURES))
        else:
            cap_list.append({})
    return ms_list, tok_list, cap_list

# ----------------------------
# Benchmark / report
# ----------------------------
@dataclass
class Result:
    name: str
    baseline_ms: List[float]
    baseline_toks: List[int]
    instr_ms: List[float]
    instr_toks: List[int]

def summarize(ms_list, tok_list):
    total_ms = sum(ms_list)
    total_s = total_ms / 1000.0
    total_toks = sum(tok_list)
    tps = total_toks / total_s if total_s > 0 else float("nan")
    return {
        "p50_ms": percentile(ms_list, 50),
        "p95_ms": percentile(ms_list, 95),
        "mean_ms": float(statistics.mean(ms_list)) if ms_list else float("nan"),
        "tokens_per_sec": float(tps),
        "total_prompts": len(ms_list),
        "total_tokens": int(total_toks),
    }

def print_report(res: Result):
    base = summarize(res.baseline_ms, res.baseline_toks)
    inst = summarize(res.instr_ms, res.instr_toks)
    overhead = (inst["mean_ms"] - base["mean_ms"]) / base["mean_ms"] * 100.0 if base["mean_ms"] > 0 else float("nan")

    print("\n" + "="*80)
    print(f"{res.name}")
    print("-"*80)
    print("Baseline (no QKV hooks):")
    print(json.dumps(base, indent=2))
    print("Instrumented (QKV extraction on):")
    print(json.dumps(inst, indent=2))
    print(f"Overhead (mean_ms): {overhead:.2f}%")
    print("="*80 + "\n")

# ----------------------------
# Main
# ----------------------------
ensure_hf_login()

# Load prompts
mt_path = OUT_DIR / "mt_bench_question.jsonl"
download_mtbench(mt_path)
prompts = load_mtbench_first_turns(mt_path, MAX_PROMPTS)
print(f"Loaded {len(prompts)} prompts from MT-Bench (first turns).")

# Warmup subset
warm_prompts = prompts[:min(WARMUP, len(prompts))]
test_prompts = prompts[min(WARMUP, len(prompts)):]  # measured prompts

# ---- Hugging Face ----
print("\nLoading Hugging Face model...")
hf_tok, hf_model = hf_load()

HF_CAPTURES: Dict[int, Dict] = {}
hf_handles = hf_attach_qkv_hooks(
    hf_model,
    LAYERS_TO_CAPTURE,
    HEADS_TO_CAPTURE,
    TOKEN_POSITIONS,
    HF_CAPTURES,
    save_raw_for_first_n=SAVE_RAW_SLICES_FOR_FIRST_N,
)

# Baseline: remove hooks temporarily for true baseline
for h in hf_handles:
    h.remove()

print("HF warmup (baseline)...")
_ = hf_prefill(hf_tok, hf_model, warm_prompts, capture=False)

print("HF baseline run...")
hf_base_ms, hf_base_toks, _ = hf_prefill(hf_tok, hf_model, test_prompts, capture=False)

# Instrumented
HF_CAPTURES.clear()
hf_handles = hf_attach_qkv_hooks(
    hf_model,
    LAYERS_TO_CAPTURE,
    HEADS_TO_CAPTURE,
    TOKEN_POSITIONS,
    HF_CAPTURES,
    save_raw_for_first_n=SAVE_RAW_SLICES_FOR_FIRST_N,
)

print("HF warmup (instrumented)...")
_ = hf_prefill(hf_tok, hf_model, warm_prompts, capture=True)

print("HF instrumented run...")
hf_inst_ms, hf_inst_toks, hf_inst_caps = hf_prefill(hf_tok, hf_model, test_prompts, capture=True)

# Save a small capture snapshot for debugging
with open(OUT_DIR / "hf_capture_example.json", "w") as f:
    json.dump(hf_inst_caps[0] if hf_inst_caps else {}, f, indent=2)

for h in hf_handles:
    h.remove()

hf_result = Result(
    name="Hugging Face Transformers",
    baseline_ms=hf_base_ms,
    baseline_toks=hf_base_toks,
    instr_ms=hf_inst_ms,
    instr_toks=hf_inst_toks,
)

# ---- LitGPT ----
print("\nLoading LitGPT model...")
from litgpt import LLM  # triggers import errors early if any
lit_llm = litgpt_load()

# Baseline: no hooks
print("LitGPT warmup (baseline)...")
_ = litgpt_prefill(lit_llm, warm_prompts, capture=False)

print("LitGPT baseline run...")
lit_base_ms, lit_base_toks, _ = litgpt_prefill(lit_llm, test_prompts, capture=False)

# Instrumented
LIT_CAPTURES: Dict[int, Dict] = {}
lit_handles = litgpt_attach_qkv_hooks(
    lit_llm,
    LAYERS_TO_CAPTURE,
    HEADS_TO_CAPTURE,
    TOKEN_POSITIONS,
    LIT_CAPTURES,
    save_raw_for_first_n=SAVE_RAW_SLICES_FOR_FIRST_N,
)

print("LitGPT warmup (instrumented)...")
_ = litgpt_prefill(lit_llm, warm_prompts, capture=True)

print("LitGPT instrumented run...")
lit_inst_ms, lit_inst_toks, lit_inst_caps = litgpt_prefill(lit_llm, test_prompts, capture=True)

with open(OUT_DIR / "litgpt_capture_example.json", "w") as f:
    json.dump(lit_inst_caps[0] if lit_inst_caps else {}, f, indent=2)

for h in lit_handles:
    h.remove()

lit_result = Result(
    name="LitGPT",
    baseline_ms=lit_base_ms,
    baseline_toks=lit_base_toks,
    instr_ms=lit_inst_ms,
    instr_toks=lit_inst_toks,
)

# ---- Report ----
print_report(hf_result)
print_report(lit_result)

# Save summary JSON
summary = {
    "model": MODEL_ID,
    "dtype": str(DTYPE),
    "device": DEVICE,
    "truncate_to_tokens": TRUNCATE_TO_TOKENS,
    "layers_to_capture": LAYERS_TO_CAPTURE,
    "heads_to_capture": HEADS_TO_CAPTURE,
    "token_positions": TOKEN_POSITIONS,
    "hf": {
        "baseline": summarize(hf_result.baseline_ms, hf_result.baseline_toks),
        "instrumented": summarize(hf_result.instr_ms, hf_result.instr_toks),
    },
    "litgpt": {
        "baseline": summarize(lit_result.baseline_ms, lit_result.baseline_toks),
        "instrumented": summarize(lit_result.instr_ms, lit_result.instr_toks),
    },
}
with open(OUT_DIR / "summary.json", "w") as f:
    json.dump(summary, f, indent=2)

print(f"\nDone. Outputs written to: {OUT_DIR.resolve()}")
print("Key files:")
print(f"  - {OUT_DIR/'summary.json'}")
print(f"  - {OUT_DIR/'hf_capture_example.json'}")
print(f"  - {OUT_DIR/'litgpt_capture_example.json'}")
print(f"  - raw slice NPZ files (if enabled), e.g. hf_qkv_layer*_sample*.npz")
```
